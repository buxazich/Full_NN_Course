### Матрично и векторное исчисление.

```python
import numpy as np

a = np.arange(1, 10).reshape(3,3)
b = np.arange(10, 19).reshape(3,3)
b
```

```python
np.dot(a,b)
```

```python
np.dot(b,a)
```

```python
np.matmul(a,b)
```

```python
a = np.arange(1,10)
b = np.ones(9)
b
```

```python
a 
```

```python
np.dot(a,b)
```

```python
np.inner(a,b)
```

```python
np.outer(a,b)
```

```python
np.outer(b,a)
```
