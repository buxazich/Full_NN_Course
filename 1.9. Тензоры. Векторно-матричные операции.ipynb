```python
import torch
```

```python
a = torch.arange(1,10).view(3,3)
b = torch.arange(10,19).view(3,3)
a
```

```python
b
```

```python
# Умножение тензора на тензор (Поэлементное, не матричное)
r1 = a*b
r1
```

```python
r2 = torch.mul(a,b)
r2
```

```python
# Матричное умножение
r3 = torch.matmul(a,b)  # перемножение с возможностью транслирования (матрицы на вектор)
r3
```

```python
r4 = torch.mm(a,b) # перемножение без транслирования
r4
```

```python
r3 = a.matmul(b)
r3
```
